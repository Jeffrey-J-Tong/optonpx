## 0. How the neuropixels probes work & how to prepare recording

### 0.1. How the electrodes are mapped to the channels:

We can not record from all the electrodes of a neuropixels probe. We can only map 384 electrodes to the 384 channels and acquire signal from them. The connected electrodes can not be arbitraily selected. For a single channel, only one of several fixed electrodes (3-4 for Neuropixels 2.0 1-shank, 12-16 for Neuropixels 2.0 4-shank) can be connected.
