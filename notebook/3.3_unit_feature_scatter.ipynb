# Unit feature scatter plots + putative cell type, with optotagging overlay

Reads the optotagging summary of notebook 3.2 for one probe/session, computes
waveform and firing-pattern features for every curated unit, classifies
putative SPNs, and writes `unit_classification.csv`.

Following Dan's pipeline (`add_raw_waveform_features_and_type_to_units_v2.m`):

- **Waveforms** are read from the raw (unfiltered) binary Kilosort ran on, on
  each unit's Phy peak channel (`ch`), using only **behavior-session** spikes
  (no laser-evoked spikes): up to 500 evenly spaced spikes, -1 to +2 ms.
  Features (`src.waveform.waveform_features`): peak width = full width at half
  amplitude of the main deflection (half-width), p2v = time from it to the
  opposite-polarity extremum.
- **Firing pattern** (rate, CV(ISI), Prop(ISI > threshold)) from
  behavior-session spikes only.
- **Putative SPN**: peak width > 90 us, p2v > 560 us, firing rate < 15 Hz
  (Dan's defaults, editable in `spn_params`).

Blue and red tags are recorded independently (no dual-colour policy).
Run notebook 3.2 first.

```python
from pathlib import Path
import yaml

# ======== parameters to change ========
session_name    = "IM-1979_2026-09-04"
base_dir_raw    = Path(r"D:/")                # holds <animal>/ephys/<raw session folders> -- needed for bit_volts
base_dir_concat = Path(r"E:/")
tagging_probe   = "probeB"
kilosort_dir    = "kilosort4"                 # KS4 results subfolder name under <processed_session>/<tagging_probe>/
unit_criteria   = {"quality": ["good"]}       # must match what was used in notebook 3.2, so unit_ids line up with opto_df
isi_long_thresh_s = 2.0                        # Prop(ISI > this) threshold
spn_params      = {"min_peak_width_us": 90.0,  # putative SPN: peak width > , p2v > , firing rate <
                   "min_p2v_us": 560.0,
                   "max_fr_hz": 15.0}
# ======================================

metadata_dir  = Path.cwd().parent
metadata_path = metadata_dir / "metadata" / f"{session_name}.yaml"
with open(metadata_path) as f:
    config = yaml.safe_load(f)

processed_path = base_dir_concat / config["animal"] / "ephys" / config["sessions"]["processed"]
probe_names    = config["streams"]["probes"]

if tagging_probe not in probe_names:
    raise ValueError(f"tagging_probe={tagging_probe!r} not in this session's probes {probe_names}")

print(f"processed_session = {processed_path}")
print(f"tagging_probe     = {tagging_probe}  (available: {probe_names})")
```

```python
# ── load curated sorting, optotagging summary, binary + session bounds ───────
import sys, json
sys.path.append("..")

import numpy as np
import pandas as pd
from src import load_curated_sorting, read_kilosort_params, oe_parse_folders, oe_parse_params

sorting = load_curated_sorting(processed_path, tagging_probe, kilosort_dir=kilosort_dir,
                               unit_criteria=unit_criteria)
fs = sorting.get_sampling_frequency()
print(f"{tagging_probe}: {len(sorting.unit_ids)} units kept (unit_criteria={unit_criteria})")

summary_path = processed_path / tagging_probe / "optotagging_summary.csv"
if not summary_path.exists():
    raise FileNotFoundError(f"{summary_path} not found -- run notebook 3.2 for this session first")
opto_df = pd.read_csv(summary_path)

# the binary Kilosort ran on (raw, unfiltered); opened read-only by src.waveform
kilosort_path = processed_path / tagging_probe / kilosort_dir
ks_params = read_kilosort_params(kilosort_path)
print(f"dat: {ks_params['dat_path']}  ({ks_params['n_channels_dat']} channels)")

# behavior session = [sample_start, sample_end) on the concatenated axis
with open(processed_path / "session_boundaries.json") as f:
    sessions = json.load(f)["streams"][tagging_probe]["sessions"]
behavior = next(s for s in sessions if s["session_id"] == "behavior")
behavior_interval = (int(behavior["sample_start"]), int(behavior["sample_end"]))
behavior_duration_s = (behavior_interval[1] - behavior_interval[0]) / fs
print(f"behavior session: samples {behavior_interval}, {behavior_duration_s:.1f} s")

# bit_volts is a per-probe value read from the raw session's settings.xml/oebin, not a
# fixed hardware constant -- any raw session for this probe carries the same value.
raw_session_dir = base_dir_raw / config["animal"] / "ephys" / config["sessions"]["optotagging"]
_, raw_oe_paths = oe_parse_folders(raw_session_dir)
raw_probe_params, _ = oe_parse_params(raw_oe_paths["xml"], raw_oe_paths["oebin"])
bit_volts = float(next(p for p in raw_probe_params
                       if str(p["stream_name"]).split(".")[-1].lower() == tagging_probe.lower())
                  ["channel_bit_volts"])
print(f"bit_volts = {bit_volts}")
```

```python
# ── one row per unit: raw-waveform features + behavior-session firing pattern ──
from src import compute_unit_waveform_features

waveform_df, unit_waveforms = compute_unit_waveform_features(
    sorting, ks_params["dat_path"], ks_params["n_channels_dat"],
    spike_interval=behavior_interval, bit_volts=bit_volts, return_waveforms=True)

rows = []
for unit_id in sorted(sorting.unit_ids):
    spike_samples = sorting.get_unit_spike_train(unit_id)
    in_behavior = (spike_samples >= behavior_interval[0]) & (spike_samples < behavior_interval[1])
    isi = np.diff(spike_samples[in_behavior] / fs)
    rows.append({
        "unit_id": unit_id,
        "n_spikes_behavior": int(in_behavior.sum()),
        "firing_rate_hz": in_behavior.sum() / behavior_duration_s,
        "cv_isi": isi.std() / isi.mean() if len(isi) > 1 else np.nan,
        "prop_isi_long": (isi > isi_long_thresh_s).mean() if len(isi) else np.nan,
    })
firing_df = pd.DataFrame(rows).set_index("unit_id")

features_df = firing_df.join(waveform_df)
n_failed = int(features_df["feature_failed"].sum())
if n_failed:
    print(f"NOTE: waveform features failed for {n_failed} units:")
    print(features_df.loc[features_df["feature_failed"], "fail_reason"].value_counts())

print(f"{len(features_df)} units in features_df")
features_df[["firing_rate_hz", "cv_isi", "prop_isi_long", "peak_width_us", "p2v_us"]].describe()
```

```python
# ── optotagging per colour + putative SPN classification ─────────────────────
# color_tagged is True per (unit_id, color) if that unit is tagged at >= 1 power
# level of that color; blue and red are recorded independently (no dual-colour policy).
from src import classify_putative_spn

tagged_blue = set(opto_df.loc[(opto_df["color"] == "blue_laser") & opto_df["color_tagged"], "unit_id"])
tagged_red  = set(opto_df.loc[(opto_df["color"] == "red_laser")  & opto_df["color_tagged"], "unit_id"])
features_df["blue_tagged"] = [u in tagged_blue for u in features_df.index]
features_df["red_tagged"]  = [u in tagged_red for u in features_df.index]

def tag_label(unit_id):
    in_blue, in_red = unit_id in tagged_blue, unit_id in tagged_red
    if in_blue and in_red:
        return "both"
    if in_blue:
        return "blue"
    if in_red:
        return "red"
    return "none"

features_df["tag"] = [tag_label(u) for u in features_df.index]
features_df = classify_putative_spn(features_df, params=spn_params)

TAG_COLORS = {"none": "0.75", "blue": "tab:blue", "red": "tab:red", "both": "tab:purple"}
TAG_ORDER  = ["none", "blue", "red", "both"]   # plot 'none' first so it sits behind the tagged points

print(pd.crosstab(features_df["tag"], features_df["is_spn"].map({True: "SPN", False: "not SPN"}),
                  margins=True).reindex(TAG_ORDER + ["All"], fill_value=0))
print("\nnot-SPN reasons:")
print(features_df.loc[~features_df["is_spn"], "spn_fail_reason"].value_counts())
```

```python
# ── Figure 1: peak width vs peak-to-valley (raw waveforms), SPN gate dashed ────
%matplotlib inline
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(5, 4.5))
for tag in TAG_ORDER:
    d = features_df[features_df["tag"] == tag]
    ax.scatter(d["p2v_us"], d["peak_width_us"],
               s=28 if tag != "none" else 14, c=TAG_COLORS[tag], label=f"{tag} (n={len(d)})",
               alpha=0.85 if tag != "none" else 0.6)
ax.axvline(spn_params["min_p2v_us"], color="k", ls="--", lw=0.8)
ax.axhline(spn_params["min_peak_width_us"], color="k", ls="--", lw=0.8)
ax.set_xlabel("peak-to-valley (us)")
ax.set_ylabel("peak width / half-width (us)")
ax.set_title(f"{session_name}  {tagging_probe}")
ax.legend()
fig.tight_layout()
```

```python
# ── Figure 2: firing rate / Prop(ISI > threshold) / CV(ISI) (3D) ────────────
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401 -- registers the '3d' projection

fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(projection="3d")
for tag in TAG_ORDER:
    d = features_df[features_df["tag"] == tag]
    ax.scatter(d["firing_rate_hz"], d["prop_isi_long"], d["cv_isi"],
               s=28 if tag != "none" else 14, c=TAG_COLORS[tag], label=f"{tag} (n={len(d)})",
               alpha=0.85 if tag != "none" else 0.6)
ax.set_xlabel("firing rate (Hz)")
ax.set_ylabel(f"Prop(ISI > {isi_long_thresh_s}s)")
ax.set_zlabel("CV(ISI)")
ax.set_title(f"{session_name}  {tagging_probe}")
ax.legend()
fig.tight_layout()
```

## Per-unit diagnostic plot

For one unit at a time: raw mean waveform split by baseline / blue-laser /
red-laser spikes (on the Phy peak channel, same reader as the classification),
its position on the probe, and where it sits in the peak-to-valley vs
half-width population. This complements notebook 3.2's `plot_opto_response`
(raster/PSTH).

```python
# ── load laser events + probe geometry for the diagnostic plot ──────────────
from src import load_laser_events, load_laser_protocol, annotate_laser_events_with_protocol
from src import open_raw_binary
from src.optotagging import DEFAULT_OPTOTAG_PARAMS

laser_events = load_laser_events(processed_path, probe_name=tagging_probe, session_id="optotagging")
protocol = load_laser_protocol(processed_path / "laser_control_protocol_dual_mega2560.csv")
laser_events = annotate_laser_events_with_protocol(laser_events, protocol)

# Same response window compute_population_opto_response used to decide tagging in
# 3.2 -- keep the baseline/blue/red split consistent with how tagging itself works.
# response window = [response_start_ms, pulse width] (pulse width from the protocol csv)
resp_start_ms = DEFAULT_OPTOTAG_PARAMS["response_start_ms"]
pulse_widths = laser_events["pulse_width_ms"].unique()
assert len(pulse_widths) == 1, f"expected one pulse width, got {pulse_widths}"
resp_end_ms = float(pulse_widths[0])
print(f"response window: [{resp_start_ms}, {resp_end_ms}] ms")

# Kilosort channel geometry: row i of channel_positions belongs to dat channel channel_map[i]
channel_map       = np.load(kilosort_path / "channel_map.npy").ravel()
channel_positions = np.load(kilosort_path / "channel_positions.npy")
raw_dat = open_raw_binary(ks_params["dat_path"], ks_params["n_channels_dat"])   # read-only memmap
print(f"laser_events: {len(laser_events)} pulses, colors={sorted(laser_events['color'].unique())}")
```

```python
# ── spike classification by laser condition ─────────────────────────────────
def classify_spikes_by_laser(spike_samples_concat, laser_events, fs,
                             resp_start_ms=resp_start_ms, resp_end_ms=resp_end_ms):
    """dict: color -> bool mask over spike_samples_concat (spike falls in the
    response window after >=1 pulse of that color), plus 'baseline' = not in
    any color's window."""
    spike_s = spike_samples_concat / fs
    masks = {}
    any_evoked = np.zeros(len(spike_s), dtype=bool)
    for color in laser_events["color"].unique():
        pulses_s = laser_events.loc[laser_events["color"] == color, "on_sample_concat"].to_numpy() / fs
        in_window = np.zeros(len(spike_s), dtype=bool)
        for t0 in pulses_s:
            in_window |= (spike_s >= t0 + resp_start_ms / 1000) & (spike_s <= t0 + resp_end_ms / 1000)
        masks[color] = in_window
        any_evoked |= in_window
    masks["baseline"] = ~any_evoked
    return masks
```

```python
# ── plot_unit_diagnostic: raw waveform-by-condition + probe position + p2v/half-width ──
#   diagnostic_figures_mode == "plot" : plot n_plot_preview random tagged units inline
#   diagnostic_figures_mode == "save" : save every tagged unit to disk, format set by figure_format
#       figure_format == "pdf" : all units -> one multi-page unit_diagnostic_figures.pdf
#       figure_format == "svg" / "png" : one file per unit, in unit_diagnostic_figures_<format>/
#                                         (png saved with a transparent background, 300 dpi)
import random
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
from tqdm.auto import tqdm
from src import mean_raw_waveform
from src.waveform import evenly_spaced


def plot_unit_diagnostic(unit_id, ms_before=1.5, ms_after=2.5, max_spikes=500):
    row = features_df.loc[unit_id]
    peak_ch = int(row["ch"])
    spike_samples = sorting.get_unit_spike_train(unit_id).astype(np.int64)
    masks = classify_spikes_by_laser(spike_samples, laser_events, fs)
    n_before = int(round(ms_before / 1000 * fs))
    n_after  = int(round(ms_after / 1000 * fs))

    fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

    # panel A: raw mean waveform by condition on the unit's Phy peak channel
    wf_colors = {"baseline": "0.3", "blue_laser": "tab:blue", "red_laser": "tab:red"}
    for cond, color in wf_colors.items():
        mask = masks.get(cond)
        if mask is None or mask.sum() == 0:
            continue
        wf, n = mean_raw_waveform(raw_dat, evenly_spaced(spike_samples[mask], max_spikes),
                                  peak_ch, n_before, n_after, bit_volts)
        if wf is None:
            continue
        t_ms = (np.arange(len(wf)) - n_before) / fs * 1000
        axes[0].plot(t_ms, wf - np.median(wf[:n_before // 2]), color=color, label=f"{cond} (n={n})")
    axes[0].axvline(0, color="0.85", lw=0.8, zorder=0)
    axes[0].set_xlabel("time (ms)")
    axes[0].set_ylabel("amplitude (uV, baseline-subtracted)")
    axes[0].set_title(f"raw waveform (Phy ch {peak_ch})")
    axes[0].legend(fontsize=8)

    # panel B: probe position, this unit's peak channel highlighted
    axes[1].scatter(channel_positions[:, 0], channel_positions[:, 1], s=8, c="0.8")
    highlight_color = TAG_COLORS.get(row["tag"], "k")
    ch_rows = np.flatnonzero(channel_map == peak_ch)
    if ch_rows.size:
        axes[1].scatter(*channel_positions[ch_rows[0]], s=80, c=highlight_color, edgecolor="k", linewidth=0.5)
    axes[1].set_xlabel("x (um)")
    axes[1].set_ylabel("y (um)")
    axes[1].set_title("probe position")
    axes[1].set_aspect("equal")

    # panel C: peak-to-valley vs half-width, only this unit colored
    axes[2].scatter(features_df["p2v_us"], features_df["peak_width_us"], s=10, c="0.8")
    axes[2].scatter(row["p2v_us"], row["peak_width_us"], s=90, c=highlight_color, edgecolor="k", linewidth=0.5)
    axes[2].axvline(spn_params["min_p2v_us"], color="k", ls="--", lw=0.8)
    axes[2].axhline(spn_params["min_peak_width_us"], color="k", ls="--", lw=0.8)
    axes[2].set_xlabel("peak-to-valley (us)")
    axes[2].set_ylabel("peak width / half-width (us)")
    axes[2].set_title("waveform-shape population")

    spn_text = "SPN" if row["is_spn"] else f"not SPN ({row['spn_fail_reason']})"
    fig.suptitle(f"unit {unit_id}  (tag={row['tag']}, {spn_text}, "
                 f"behavior firing rate={row['firing_rate_hz']:.2f} Hz)")
    fig.tight_layout()
    return fig


# ======== parameters to change ========
diagnostic_figures_mode = "save"   # "plot" or "save"
figure_format = "pdf"              # "pdf", "svg", or "png" -- only used when mode == "save"
n_plot_preview = 4                 # number of random tagged units to preview when mode == "plot"
# ======================================

tagged_unit_ids = sorted(tagged_units_df.index.tolist())
print(f"{len(tagged_unit_ids)} tagged units")

if diagnostic_figures_mode == "plot":
    for unit_id in sorted(random.sample(tagged_unit_ids, k=min(n_plot_preview, len(tagged_unit_ids)))):
        plot_unit_diagnostic(unit_id)

elif diagnostic_figures_mode == "save":
    if figure_format == "pdf":
        pdf_path = processed_path / tagging_probe / "unit_diagnostic_figures.pdf"
        with PdfPages(pdf_path) as pdf:
            for unit_id in tqdm(tagged_unit_ids, desc="pdf pages", unit="unit"):
                fig = plot_unit_diagnostic(unit_id)
                pdf.savefig(fig, bbox_inches="tight")
                plt.close(fig)
        print(f"saved {len(tagged_unit_ids)} pages to {pdf_path}")

    elif figure_format in ("svg", "png"):
        out_dir = processed_path / tagging_probe / f"unit_diagnostic_figures_{figure_format}"
        out_dir.mkdir(exist_ok=True)
        for unit_id in tqdm(tagged_unit_ids, desc=f"{figure_format} files", unit="unit"):
            fig = plot_unit_diagnostic(unit_id)
            fig_path = out_dir / f"unit_{unit_id}.{figure_format}"
            if figure_format == "png":
                fig.savefig(fig_path, bbox_inches="tight", transparent=True, dpi=300)
            else:
                fig.savefig(fig_path, bbox_inches="tight", transparent=True)
            plt.close(fig)
        print(f"saved {len(tagged_unit_ids)} files to {out_dir}")

    else:
        raise ValueError(f"figure_format must be 'pdf', 'svg', or 'png', got {figure_format!r}")

else:
    raise ValueError(f"diagnostic_figures_mode must be 'plot' or 'save', got {diagnostic_figures_mode!r}")
```
